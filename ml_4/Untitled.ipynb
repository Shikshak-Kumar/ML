# 1. Convert everyting to lower case 

```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

# 2. Remove punctuation (not req in ML can be used in DL)

```python
import string 

def remove_punc(txt): 
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punc)
```

# 3. Remove numbers

```python
def remove_nums(txt): 
    new = ""
    for i in txt:
        if not i.isdigit(): 
            new = new + i
    return new
```

```python
df['text'] = df['text'].apply(remove_nums)
```

# 6. Remove Emojis & Special characters

```python
def remove_emojis(txt): 
    new = ""
    for i in txt:
        if i.isascii(): 
            new = new + i
    return new

df['text'] = df['text'].apply(remove_emojis)
```

# 7. Remove stop keywords: "is", "was", "and", "in" (not req in ML can be used in DL)

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
len(stop_words)
```

```python
df.loc[1]['text']
```

```python
def remove_stopwords(txt): 
    words = txt.split() # do this if word_tokenize not working
    cleaned = []
    for i in words: 
        if not i in stop_words: 
            cleaned.append(i)
    return ' '.join(cleaned)
```

```python
df['text'] = df['text'].apply(remove_stopwords)
```
