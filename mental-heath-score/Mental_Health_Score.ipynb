# Checking Outliers

## Need to calculate IQR (Interquartile Range) for checking outliers- for that we need Q1,Q3

## Quartiles (Q1, Q2, and Q3)

Quartiles divide a sorted dataset into four equal parts and help us understand the distribution of the data.

- **Q1 (First Quartile):** Represents the 25th percentile. It is the value below which 25% of the data falls.
- **Q2 (Second Quartile / Median):** Represents the 50th percentile. It divides the dataset into two equal halves.
- **Q3 (Third Quartile):** Represents the 75th percentile. It is the value below which 75% of the data falls.

For a dataset containing `N` values, the positions of the quartiles can be calculated as:

- **Q1 Position:** `(N + 1) / 4`
- **Q2 Position:** `2(N + 1) / 4`
- **Q3 Position:** `3(N + 1) / 4`

Quartiles are commonly used in Exploratory Data Analysis (EDA) to understand data distribution and identify potential outliers using the Interquartile Range (IQR).

### Example

Sorted Data:

20, 20, 22, **22**, 22, 24, 25, **27**, 30, 30, 34, **35**, 35, 36, 40

Here, N = 15

- Q1 Position = (15 + 1) / 4 = 4th term = 22
- Q2 Position = 2(15 + 1) / 4 = 8th term = 27
- Q3 Position = 3(15 + 1) / 4 = 12th term = 35

```python
num_features = df.select_dtypes(include='number') # int64, float64 

Q1 = num_features.quantile(0.25)
Q3 = num_features.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5*IQR 
upper_bound = Q3 + 1.5*IQR 

outliers = (num_features < lower_bound) | (num_features > upper_bound)

print(outliers.sum())
```

## Outlier Detection using IQR

Before detecting outliers, only numerical columns are selected from the dataset. The IQR method is then applied to each numerical feature independently.

`Q1` represents the 25th percentile and `Q3` represents the 75th percentile of each numerical column. The Interquartile Range (IQR) is calculated as the difference between Q3 and Q1.

Lower and upper bounds are then calculated using the IQR method. Any value below the lower bound or above the upper bound is considered a potential outlier.

The calculation is performed column-wise, allowing outliers to be detected separately for every numerical feature.

# How the IQR Outlier Calculation Works

## 1. Calculate Q1 for every numerical column
    Q1 = num_features.quantile(0.25)

Returns something like:
    Age             22.5
    Screen_Time      5.85
    Mental_Score    52.5

Each column gets its own Q1.

## 2. Calculate Q3 for every column
    Q3 = num_features.quantile(0.75)

Example:
    Age             27.5
    Screen_Time      8.25
    Mental_Score    67.5

## 3. Calculate IQR
    IQR = Q3 - Q1

Pandas automatically matches values by column name:
    Age             5
    Screen_Time     2.4
    Mental_Score    15

## 4. Calculate lower and upper bounds
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

Each column gets separate bounds.

**Age:**
    Lower Bound = Q1 - 1.5 × IQR
    Upper Bound = Q3 + 1.5 × IQR

**Screen_Time:**
    Lower Bound = Q1 - 1.5 × IQR
    Upper Bound = Q3 + 1.5 × IQR

## 5. Detect outliers
    outliers = (num_features < lower_bound) | (num_features > upper_bound)

This might look confusing because:
- `num_features` → DataFrame (many rows and columns)
- `lower_bound` → Series (one value per column)

Pandas automatically aligns the bounds with their respective column names and compares every value in that column.

Conceptually, for every row:
    num_features['Age'] < lower_bound['Age']
    num_features['Screen_Time'] < lower_bound['Screen_Time']

The result is a DataFrame of True/False:

| Age   | Screen_Time | Mental_Score |
|-------|-------------|--------------|
| False | False       | False        |
| False | False       | False        |
| False | True        | False        |

`True` means that value is an outlier (for that specific column).
